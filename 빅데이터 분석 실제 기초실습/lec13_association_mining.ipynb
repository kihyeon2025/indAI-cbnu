## 1. 장바구니 데이터 형태

트랜잭션 데이터(0/1 인코딩) 예시

```
transactions = [
    ["milk", "bread", "eggs"],
    ["beer", "diapers", "chips"],
    ["milk", "diapers", "beer", "bread"],
    ["bread", "butter"],
    ["milk", "bread", "butter"],
]
items = sorted({i for t in transactions for i in t})
df = pd.DataFrame([{item: (item in t) for item in items} for t in transactions])
df
```

## 2. Apriori (옵션)

mlxtend로 frequent itemsets / rules 생성 (설치 필요)

```
# !pip install -q mlxtend
from mlxtend.frequent_patterns import apriori, association_rules

freq = apriori(df, min_support=0.4, use_colnames=True)
rules = association_rules(freq, metric="confidence", min_threshold=0.6)

print(freq.sort_values("support", ascending=False))
print(rules[["antecedents","consequents","support","confidence","lift"]].sort_values("lift", ascending=False))
```
