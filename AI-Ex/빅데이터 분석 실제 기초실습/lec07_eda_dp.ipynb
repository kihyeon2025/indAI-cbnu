## 1. EDA 기본 체크리스트

shape/info/describe, 결측치/중복/분포를 점검합니다.

```python
df = pd.DataFrame({
    "x": np.random.randn(1000),
    "y": np.random.randn(1000) * 2 + 1,
    "cat": np.random.choice(["A","B","C"], size=1000),
})
print(df.shape)
print(df.info())
print(df.describe())
print(df.isna().sum())
```

## 2. 이상치 탐지

IQR 기반 이상치 탐지를 실습합니다.

```python
q1, q3 = df["y"].quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
outliers = df[(df["y"] < lower) | (df["y"] > upper)]
print("outliers:", len(outliers))
```

## 3. 인코딩/스케일링

One-hot encoding, 표준화/정규화를 적용합니다.

```python
dfrom sklearn.preprocessing import OneHotEncoder, StandardScaler

X_num = df[["x","y"]].values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_num)

enc = OneHotEncoder(sparse_output=False)
X_cat = enc.fit_transform(df[["cat"]])

X_all = np.hstack([X_scaled, X_cat])
print("X_all shape:", X_all.shape)
```
